# 09 · Results summary: the investigation, gate by gate

**Investigation ladder** (see [`EVALUATION.md`](../EVALUATION.md)): this notebook collects the **Result (gate)** cell of
every stage into one argument. It has no code to write.

**How this notebook works**
- One section per stage, in ladder order. Copy the gate table from that stage's notebook, paste its key figure,
  and write one paragraph: what the number means, and **where the prediction was wrong**.
- A stage only counts if its gate was checked with the shared protocol: independent unit = **recording block**
  (15 blocks), leave-one-stage-out folds, baselines and nulls, confidence intervals over blocks.
- If a gate fails, say so and stop the chain there: later stages that depend on it are reported as "not run" or
  "exploratory".

**Data:** one participant (emg2pose mini), 30 recordings in 15 simultaneous left/right blocks.

## What this means for H0 and H1
- **H0 (physiological synergies can be identified):** supported only as far as stages 1–3 pass, *and* stages 4–5 show
  the synergies matter for movement. Surface EMG alone can't prove synergies are neural (Kutch & Valero-Cuevas 2012),
  so word the claim as "necessary conditions met in one person".
- **H1 (bottlenecks reduce calibration on new users):** one person can't test it. Stages 6 and 5b's within-person
  calibration curves shows only whether the bottlenecks are **worth testing** at full scale (stage 7).

## Limitations
- One participant; no held-out users.
- 15 independent blocks; the left and right wrists are recorded at the same time, and in two blocks the idle hand
  is labelled `unconstrained`.
- Linear and state-space decoders, not the proposal's deep decoder; the pretrained vemg2pose model (eda/00) is the
  ceiling we have not yet compared with.
- Zero-phase envelopes include a little future EMG; "causal" or forecasting claims need causal features.
- Pose comes from camera tracking, with dropouts and jitter.
- The eda/03 effect sizes are pseudo-replicated and are **not** reported here.
